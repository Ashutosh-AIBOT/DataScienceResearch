## Topic 1: Scalars, Vectors, Matrices, and Tensors — Definitions and Notation

### Scaler : 
Scaler is a singular number .

### vector :
It is the ordered list of numbmer .

### Matrix :
A Grid of rectangular number .

### Tensors :
A tensor is a generalization of vectors and matrices to any number of dimensions .

## Important Variants

For this topic, the key variants are different tensor orders.

| Object        | Rank | Example Shape    |
| ------------- | ---- | ---------------- |
| Scalar        | 0    | ()               |
| Vector        | 1    | (5,)             |
| Matrix        | 2    | (3,4)            |
| Rank-3 Tensor | 3    | (10,28,28)       |
| Rank-4 Tensor | 4    | (32,224,224,3)   |
| Rank-5 Tensor | 5    | (8,16,224,224,3) |


## History
Linear algebra developed over centuries.

Major milestones:

René Descartes (1637) — coordinate geometry .                                                    
Gottfried Leibniz (late 1600s) — early matrix ideas .                                            
Arthur Cayley (1858) — matrix algebra .                                                                                
James Joseph Sylvester (1850s) — matrix terminology .                                                                         
David Hilbert (1900s) — vector spaces and functional analysis .                                                           
Tensor calculus was developed by Gregorio Ricci-Curbastro and Tullio Levi-Civita, later used extensively by Albert Einstein in general relativity.

#### Modern deep learning frameworks (NumPy, TensorFlow, PyTorch, JAX) are fundamentally tensor computation libraries.
