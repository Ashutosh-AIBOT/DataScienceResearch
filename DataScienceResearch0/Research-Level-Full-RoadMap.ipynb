## 9.2 Deep RL (Every Major Algorithm, From Scratch)

### Value-Based Methods
- **DQN** (Mnih et al., DeepMind 2013-2015):
  - Experience replay
  - Target network
  - ε-greedy exploration
  - Convolutional Q-network
  - Atari games
  - From scratch in PyTorch
- **Double DQN** (Van Hasselt et al., DeepMind 2016): Overestimation bias
- **Dueling DQN** (Wang et al., DeepMind 2016): Value + advantage streams
- **Prioritized Experience Replay** (Schaul et al., DeepMind 2016)
- **Noisy Networks** (Fortunato et al., DeepMind 2017): Stochastic exploration
- **Distributional RL** (Bellemare et al., DeepMind 2017): C51
- **Rainbow** (Hessel et al., DeepMind 2017): All DQN improvements combined
- **QR-DQN**: Quantile regression
- **IQN**: Implicit quantile networks

### Policy Gradient Methods
- **REINFORCE** (Williams 1992) from scratch:
  - Monte Carlo policy gradient
  - Log-derivative trick derivation
  - Baseline for variance reduction
- **Actor-Critic** from scratch:
  - Policy (actor) + value (critic)
  - TD error as advantage
- **A2C** (Synchronous Advantage Actor-Critic) from scratch
- **A3C** (Mnih et al., DeepMind 2016): Asynchronous, multiple workers
- **TRPO** (Schulman et al., UC Berkeley 2015):
  - Trust region constraint
  - KL divergence bound
  - Conjugate gradient for optimization
- **PPO** (Schulman et al., OpenAI 2017):
  - Clipped surrogate objective
  - Much simpler than TRPO
  - Most widely used policy gradient method
  - From scratch in PyTorch — critical
  - Used in RLHF
- **GAE** (Generalized Advantage Estimation, Schulman et al., 2016):
  - λ-weighted advantage
  - Bias-variance tradeoff

### Continuous Action Methods
- **DDPG** (Lillicrap et al., DeepMind 2016):
  - Deterministic policy gradient
  - Actor-critic off-policy
  - Ornstein-Uhlenbeck exploration noise
- **TD3** (Fujimoto et al., McGill 2018):
  - Twin critics (reduce overestimation)
  - Delayed policy updates
  - Target policy smoothing
- **SAC** (Haarnoja et al., UC Berkeley 2018):
  - Maximum entropy RL
  - Soft Q-function
  - Temperature parameter α
  - State-of-the-art for continuous control
- **TQC** (Truncated Quantile Critics, 2020)

### Model-Based RL
- **World Models** (Ha & Schmidhuber, NNAISENSE 2018)
- **Dreamer** (Hafner et al., Google Brain 2020):

  - Imagine trajectories, optimize
  - DreamerV2, DreamerV3
- **MBPO** (Janner et al., UC Berkeley 2019)
- **MuZero** (Schrittwieser et al., DeepMind 2020):
  - Learn model of reward/value/policy without observing env state
  - No explicit reconstruction
  - Beats AlphaZero without knowing game rules

### Offline RL (Important)

- **IQL** (Kostrikov et al., UC Berkeley 2021): Implicit Q-learning

## 16.1 LLM Security (Red Teaming Level)

  - DAN (Do Anything Now) prompts
  - Role-play based bypasses

  - Base64/encoding tricks
  - Token smuggling

  - AutoDAN: automatic jailbreak generation
- **Data Poisoning**: Manipulate training data
- **Backdoor attacks**: Trigger patterns in fine-tuning
- **Model extraction**: Query to reconstruct model
- **Membership inference**: Was this data in training?
- **Privacy leakage**: Memorized training data extraction
- **Supply chain attacks**: Malicious model weights

- **Guardrails AI**:
  - Validators for inputs and outputs

  - Colang language for conversation flows

# 📦 PHASE 17 — POST-TRANSFORMER ARCHITECTURES

## 17.1 Efficient Transformers (Attention Variants)

### Quadratic Attention Problem
- Standard attention: O(N²) memory and compute with sequence length
- Different approaches to solve this

### Sparse Attention
- **Longformer** (Beltagy et al., AllenAI 2020):
  - Local windowed attention + global tokens
  - Linear complexity: O(N)
  - 4096+ token context
- **BigBird** (Zaheer et al., Google 2020):
  - Random + local + global attention
  - Theoretical justification for sparse
- **Sparse Transformer** (Child et al., OpenAI 2019):
  - Fixed strided patterns

### Linear Attention
- **Linformer** (Wang et al., Facebook 2020): Project K, V to lower dim
- **Performer** (Choromanski et al., Google 2021):
  - FAVOR+ (Fast Attention Via positive Orthogonal Random features)
  - Approximate softmax with random features

### Other Efficient Attention
- **Routing Transformer** (Roy et al., 2021): Online k-means routing
- **Reformer** (Kitaev et al., Google 2020):
  - LSH attention: hash to buckets
  - Reversible layers for memory
- **Luna** (Ma et al., Facebook 2021): Linear unified nested attention
- **FNet** (Lee-Thorp et al., Google 2021): Replace attention with FFT
- **MLP-Mixer** (Tolstikhin et al., Google 2021): MLP over patches (no attention)

## 17.2 State Space Models (Critical Research Area)

### S4 and Related
- **S4** (Gu et al., Stanford 2021):
  - Structured State Space Sequence Model
  - Continuous-time state space: h'(t) = Ah(t) + Bx(t)
  - HIPPO initialization for A matrix
  - Diagonal plus low-rank (DPLR) parametrization
  - Convolutional + recurrent dual modes
  - O(N log N) training, O(1) inference per step
  - Read paper carefully — foundational

- **S5**: Simplified S4 with parallel scan
- **DSS**: Diagonal state spaces
- **H3** (Fu et al., Stanford 2023): SSM + multiplicative gating
- **Hyena** (Poli et al., Stanford 2023): Subquadratic via long convolutions

### Mamba (Critical — Current Research Frontier)
- **Mamba** (Gu & Dao, Carnegie Mellon 2023):
  - Selective SSM: input-dependent (selective) state transitions
  - Selectivity: B, C, Δ are functions of input
  - Hardware-aware parallel scan (CUDA kernel)
  - No attention, linear complexity
  - Performance matches transformers at 3B+
  - From scratch implementation (extremely valuable)
  - Read paper multiple times
  
- **Mamba-2** (Dao & Gu, 2024):
  - SSD (State Space Duality)
  - Connects SSMs and attention
  - Structured Masked Attention (SMA)
  - Even more efficient

- **Vision Mamba** (VMamba, PlainMamba): Mamba for images
- **Jamba** (AI21 Labs): Mamba + Transformer layers interleaved
- **Zamba** (Zyphra): Mamba-based LLM
- **FalconMamba** (TII, UAE 2024): Pure Mamba LLM
- **MambaFormer**: Hybrid

### RWKV (BlinkDL)
- **RWKV** (Peng et al., 2023): RNN-Transformer hybrid
  - Linear attention as RNN
  - No quadratic complexity
  - Can be trained like transformer, run like RNN
  - RWKV-4, RWKV-5/Eagle, RWKV-6/Finch
  - Open source model weights up to 14B

### RetNet (Microsoft Research 2023)
- Parallel, recurrent, and chunkwise training modes
- "Training parallelism, inference efficiency, competitive performance"
- Retention mechanism

### Linear Transformers
- **GLA** (Gated Linear Attention, Yang et al., 2024)
- **Based** (Fu et al., Stanford 2024): Linear attention + sliding window
- **HGRN2** (2024)
- **Samba** (2024): Mamba + Sliding Window Attention

## 17.5 Attention Mechanism Evolution

### Multi-Query Attention (MQA)
- **MQA** (Shazeer, Google 2019):
  - Multiple query heads, single K and V head
  - Reduces KV cache size significantly
  - Used in PaLM, Falcon

### Grouped Query Attention (GQA)
- **GQA** (Ainslie et al., Google 2023):
  - Interpolation between MHA and MQA
  - G groups, each with one K, V
  - Used in LLaMA 2/3, Mistral
  - Better tradeoff than MQA

### Sliding Window Attention
- Local attention within window
- Used in Mistral 7B (4096 window)
- Longformer, BigBird

### Flash Attention Deep Dive
- IO-aware algorithm
- Tiling: compute in SRAM, avoid HBM reads
- Recomputation in backward pass
- Flash Attention 2 improvements
- Flash Attention 3 (H100 optimized)
- Triton implementation

### Differential Transformer (Microsoft 2024)
- Amplify relevant, suppress noise
- Two softmax attention maps subtracted
- Better hallucination reduction

### Ring Attention (2023)
- Distribute long sequences across devices
- Blockwise computation
- Linear scaling with number of devices

## 20.1 HOW TO READ RESEARCH PAPERS (Systematic Method)

### The Three-Pass Method
*(Srinivasan Keshav's paper: "How to Read a Paper" — read this first, 3 pages, CS.Cornell.edu)*

**Pass 1 (5-10 minutes):**
Title, abstract, introduction, section headings, conclusion, references scan — decide if worth reading fully

**Pass 2 (1 hour):**
Read carefully, skip proofs, note key figures, mark ununderstood references — understand main contribution

**Pass 3 (4-5 hours):**
Virtually re-implement — understand every assumption, challenge every claim, identify future work
