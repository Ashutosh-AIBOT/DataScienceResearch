## PART 1 — THE HONEST TRUTH ABOUT LEARNING THIS

Before anything else, understand this: **you cannot read everything fast AND deeply at the same time.** These are opposites. What you CAN do is read fast at the right depth for the right purpose. There are three modes of reading and you use each one differently.

**Mode 1 — Survey Reading (fast, 10-20 minutes per paper/chapter):** You are asking "what does this do and why does it matter?" You read abstract, intro, conclusion, look at figures. You do this for everything first before going deep. This gives you a map of the territory.

**Mode 2 — Understanding Reading (deep, 1-3 hours):** You read fully, pause at every equation, draw the diagram yourself, ask "why did they make this choice?" You do this only for papers that are core to what you are building right now.

**Mode 3 — Implementation Reading (deepest, full day):** You read, then close the paper, and try to reimplement the core idea from memory. Then re-read and fix what you got wrong. This is how you truly understand something. You do this only for foundational papers — maybe 20-30 papers total in your career.

The mistake almost everyone makes is trying to do Mode 3 on everything. You will burn out and cover nothing. The system below tells you exactly which mode to use for what.

## PART 2 — YOUR NOTE SYSTEM (CLEAN, MINIMAL, PERMANENT)

### The Single Folder Structure

Create one folder on your computer called `AI_BRAIN`. Inside it:

```
AI_BRAIN/
├── papers/          ← one file per paper
├── concepts/        ← one file per concept
├── projects/        ← your implementations
├── weekly_log/      ← what you did each week
└── questions/       ← things you don't understand yet
```

Use **Obsidian** (free) for all notes. It stores everything as plain markdown files. No lock-in, works offline, links between notes automatically, shows you a visual graph of how your knowledge connects. This matters because in 2 years, when you want to remember how HNSW connects to FAISS which connects to RAG, Obsidian shows you the path.

### Your Paper Note Template

Every single paper gets exactly this format, nothing more, nothing less. Discipline is more important than completeness:

```markdown
# [Paper Title]

**Authors:** 
**Year:** 
**Venue:** NeurIPS / ICML / ArXiv etc.
**Link:** 
**Code:**

## The Problem (1-2 sentences)
What problem were they solving? Write it in your own words.

## The Key Idea (2-3 sentences)
What was the insight that made this work? Not the method — the insight.

## How It Works (your own words, no copy-paste)
Write the method like you are explaining to someone. Use diagrams 
if helpful. If there is math, write the key equation and explain 
each symbol in plain English.

## Results That Matter
One or two numbers that tell you whether this works and how well.

## Limitations
What does this NOT do? When would you NOT use this?

## My Questions
Things I still don't understand. Revisit these later.

## Connections
[[concept this relates to]] [[other paper this reminds me of]]

## Status
- [ ] Survey read
- [ ] Full read  
- [ ] Implemented
```

This takes 15-30 minutes per paper when done honestly. If it takes longer, you are writing too much. The goal is not to summarize the paper — the paper already exists. The goal is to capture YOUR understanding in YOUR words so future you can recall it in 2 minutes.

## PART 3 — YOUR JUPYTER NOTEBOOK SYSTEM

Every topic gets one clean Jupyter notebook. The notebook has a specific purpose: it is your **proof of understanding**, not your scratch pad. Scratch work goes in a separate "experiments" notebook. The clean notebook is what you would show someone.

### Clean Notebook Format

```
Cell 1  — Markdown: Title, date, what paper/concept this is from, 
           one paragraph of what you will implement and why
           
Cell 2  — Markdown: "## Background" — 3-5 sentences of theory 
           in your own words

Cell 3  — Markdown: "## The Math" — key equations with 
           explanation of each variable

Cell 4  — Code: imports only

Cell 5  — Code: data loading or toy dataset creation
           (always use tiny data — see your GPU section below)

Cell 6  — Code: implement the core algorithm from scratch
           heavily commented — every 3-5 lines has a comment

Cell 7  — Code: verify it works on tiny example

Cell 8  — Markdown: "## Using the Library Version"

Cell 9  — Code: same thing using sklearn/PyTorch/etc.

Cell 10 — Code: compare results from scratch vs library

Cell 11 — Markdown: "## What I Learned" — 3 things that 
           surprised you or clarified your understanding

Cell 12 — Markdown: "## Questions Remaining" — honest about 
           what you still don't fully get
```

This structure means any notebook you open in 2 years is immediately understandable. No mysterious cells. No orphaned variables. No "what was I doing here" confusion.

## PART 5 — YOUR GTX 1650 + 8GB RAM + 512GB — HONEST ASSESSMENT

Your GTX 1650 has **4GB VRAM**. Your RAM is **8GB**. This is a real constraint and you should know exactly what it means so you do not waste time trying things that will not work.

### What Works Perfectly on Your Setup

**All of Classical ML:** Linear regression, SVM, random forest, XGBoost, LightGBM, clustering, dimensionality reduction — all of this runs entirely on CPU and RAM. No GPU needed. No limits. Run everything.

**Small Neural Networks from Scratch:** Your NumPy implementations, MLP on MNIST, small CNNs on CIFAR-10 — all fine. Train on GPU via PyTorch with `device = 'cuda'`. These fit easily in 4GB VRAM.

**Standard CNN Training:** ResNet-18 on CIFAR-10 with batch size 32-64 — works fine. VGG-16 at small batch size — works. This covers Phase 5 basics completely.

**Small Transformers:** A tiny GPT (like Karpathy's minGPT — 10M parameters) — trains fine. A BERT-tiny fine-tuning — fine. This lets you understand the architecture hands-on.

**Inference on Quantized LLMs via Ollama:** This is the big one. With 4GB VRAM and 8GB RAM, you can run models by offloading layers between GPU and CPU:
- **Phi-3 Mini (3.8B) Q4 quantized** — runs, slightly slow but usable
- **Gemma 2B Q4** — runs well
- **TinyLlama 1.1B** — runs very well
- **Qwen2 1.5B** — runs well
- You **cannot** run 7B models with only 4GB VRAM + 8GB RAM comfortably. The math: a 7B model at Q4 needs ~4GB VRAM + significant RAM for KV cache. It may load but will be very slow.

**RAG Systems:** ChromaDB, FAISS, embedding models (all-MiniLM-L6 is tiny) — all work. You can build complete RAG pipelines locally.

**LangChain + LangGraph with local models:** Works. Connect to Ollama running Phi-3 Mini. Full agent loops work.

**Stable Diffusion 1.5:** Works at 512×512. SD XL — will struggle, probably OOM. Stick to SD 1.5.

## PART 6 — YOUR WEEKLY LEARNING RHYTHM

This is how you actually make progress without burning out or losing depth.

**Monday-Tuesday-Wednesday:** Pick one concept from the roadmap. Read the relevant chapter or paper (Mode 2 or 3 depending on tier). Take your concept note in Obsidian. Start the Jupyter notebook.

**Thursday-Friday:** Finish the implementation in the notebook. Run it, break it, understand why it breaks, fix it. This is where understanding actually happens.

**Saturday:** Write a short summary (200-300 words) of what you learned. Post it somewhere — GitHub README, a private blog, anywhere. The act of writing forces you to find gaps in your understanding.

**Sunday:** Review. Open Obsidian, look at what you wrote on Monday. Does it still make sense? What questions do you have now that you didn't have then? Add to your questions file. Spend 30 minutes skimming 2-3 papers in Mode 1 to stay oriented on the broader landscape.

## PART 7 — REVISION SYSTEM

Forgetting is the enemy of a long roadmap. Use spaced repetition mentally, not with flashcard apps — flashcard apps work for facts but ML requires you to understand relationships.

Your revision system is simple: every Sunday, you also open the notes from 4 weeks ago and ask three questions. Can you explain the core idea in 2 sentences without reading the note? Can you recall why it matters? Can you draw or write the key equation from memory? If yes to all three, move on. If no, spend 20 minutes re-engaging with your notebook from that week.

Every month, open your Obsidian graph view and look at your connected notes. Find isolated notes — concepts with no connections to other concepts. These are danger zones. Either they are unimportant (and you can deprioritize), or they are important and you have not yet understood how they connect to everything else.

Every 3 months, pick one thing from Phase 3 (Classical ML) and re-implement it from scratch without looking at your old code. Foundations decay fastest because you use them least once you are doing advanced work.
