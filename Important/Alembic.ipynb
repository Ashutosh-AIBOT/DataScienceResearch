## Alembic Commands

**Setup**
- `pip install alembic`
- `alembic init migrations`

**Everyday**
- `alembic current` — show DB version
- `al`membic history` — show all migrations
- `alembic heads` — show latest migration(s)
- `alembic upgrade head` — apply all pending migrations
- `alembic upgrade <revision>` — upgrade to specific version
- `alembic downgrade -1` — undo last migration
- `alembic downgrade base` — undo all migrations
- `alembic check` — check if models match DB

**New migration**
1. Edit models file
2. `alembic revision --autogenerate -m "describe change"`
3. Review the generated file in `migrations/versions/`
4. `alembic upgrade head`
5. `alembic current` (confirm)

**Manual migration**
- `alembic revision -m "description"` (write upgrade/downgrade by hand)

**Stamping**
- `alembic stamp head` — mark DB as current without running migrations
- `alembic stamp <revision>` — mark DB at specific version

**Preview only**
- `alembic upgrade head --sql`

**Override DB URL**
- `DATABASE_URL="postgresql://user:pass@host:5432/db" alembic upgrade head`

**Cheat sheet**
| Situation | Command |
|---|---|
| New dev clones repo | `alembic upgrade head` |
| Changed a model | `alembic revision --autogenerate -m "..."` → `alembic upgrade head` |
| Undo last change | `alembic downgrade -1` |
| DB has tables, no history | `alembic stamp head` |
| Check DB version | `alembic current` |
| Preview before running | `alembic upgrade head --sql` |
